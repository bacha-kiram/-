```python
#📚 Workbook practice 05
def filter_even_numbers(numbers):
  result=[]
  for num in numbers :
    if num % 2 == 0 :
      result.append(num)
  return result
test = [1, 2, 3, 4, 5, 6, 7, 8]
print(filter_even_numbers(test))
```

```python
#📚 Workbook practice 06
# تعريف بالقوائم
students = ["Alice", "Bob", "Charlie", "David", "Eve"]
grades = [85, 90, 78, 92, 90]

# إزالة العلامات المكررة
unique_grades = set(grades)

# دالة إيجاد أعلى علامة

highest_grade = max(grades)

# دالة إنشاء قاموس يربط الطلاب بعلاماتهم
student_grades = dict(zip(students, grades))

# استخراج الطلاب الحاصلين على أعلى علامة
top_students = [student for student, grade in student_grades.items() if grade == highest_grade]

# طباعة النتائج
print("Student Grades Dictionary:", student_grades)
print("Unique Grades Set:", unique_grades)
print("Highest Grade:", highest_grade)
print("Top Student(s):", top_students)
```

```python
#📚 Workbook practice 07
def create_dict_from_lists(keys, values):
    if len(keys) != len(values):
        return "Error: Lists must have the same length."

    return dict(zip(keys, values))


keys_list = ["name", "age", "city"]
values_list = ["Alice", 25, "New York"]

generated_dict = create_dict_from_lists(
    keys_list, values_list
)

print("Generated Dictionary:", generated_dict)
```
